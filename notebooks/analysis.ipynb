```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
df = pd.read_csv('../data/arctic_ice_extent.csv')
df.columns = df.columns.str.strip()
df['extent'] = df['extent'].astype(str).str.strip().astype(float)
print(df.shape)
df.head(10)
```

```python
print('--- Dataset Info ---')
df.info()
print('\n--- Missing Values ---')
print(df.isnull().sum())
print('\n--- Descriptive Statistics ---')
df.describe()
```

```python
# Feature Engineering
df['decade'] = (df['year'] // 10 * 10).astype(str) + 's'
df['change_from_prev'] = df['extent'].diff()
df['rolling_avg_5yr'] = df['extent'].rolling(5, min_periods=1).mean()
df['anomaly'] = df['extent'] - df['extent'].mean()
df['era'] = pd.cut(df['year'], bins=[1978,1989,1999,2009,2030],
                   labels=['Pre-1990','1990s','2000s','2010s+'])
df.head()
```

```python
# Line Chart
fig, ax = plt.subplots(figsize=(10,4))
ax.plot(df['year'], df['extent'], marker='o', ms=4, label='Extent')
ax.plot(df['year'], df['rolling_avg_5yr'], '--', label='5-yr Avg')
ax.set_title('Arctic Sea Ice Extent Over Time')
ax.set_xlabel('Year'); ax.set_ylabel('Extent (million km²)')
ax.legend(); plt.tight_layout(); plt.show()
```

```python
# Correlation Heatmap
num_cols = ['year','extent','change_from_prev','rolling_avg_5yr','anomaly']
fig, ax = plt.subplots(figsize=(7,5))
sns.heatmap(df[num_cols].dropna().corr(), annot=True, fmt='.2f', cmap='coolwarm', ax=ax)
ax.set_title('Correlation Heatmap')
plt.tight_layout(); plt.show()
```

```python
# Box Plot by Era
fig, ax = plt.subplots(figsize=(7,4))
df.boxplot(column='extent', by='era', ax=ax)
ax.set_title('Sea Ice Extent by Era')
ax.set_xlabel('Era'); ax.set_ylabel('Extent (million km²)')
plt.suptitle('')
plt.tight_layout(); plt.show()
```
